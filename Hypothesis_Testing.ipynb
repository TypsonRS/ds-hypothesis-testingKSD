A company claims that the average delivery time is 30 minutes.

A random sample of 64 deliveries has an average delivery time of 32 minutes. The population standard deviation is known to be 8 minutes.

At a 5% significance level, test whether the actual average delivery time is different from 30 minutes.

```python
import math
from scipy.stats import norm

# -----------------------------------------
# 1. Given data
# -----------------------------------------

sample_mean = 32          # Mean of our sample
population_mean = 30      # Mean assumed under H0
population_std = 8        # Population standard deviation (known)
sample_size = 64          # Number of observations in the sample
alpha = 0.05              # Significance level


# -----------------------------------------
# 2. Calculate the Z-statistic
# -----------------------------------------

z = (sample_mean - population_mean) / (
    population_std / math.sqrt(sample_size)
)

print("Z-statistic:", z)


# -----------------------------------------
# 3. Calculate the p-value
# -----------------------------------------

# abs(z) gives the absolute value of z.
# Here:
# z = 2
# abs(2) = 2

z_absolute = abs(z)

# norm.cdf(z_absolute) gives the cumulative
# probability to the LEFT of z = 2
#
# For z = 2:
# CDF(2) ≈ 0.9772
#
# This means approximately 97.72% of the
# standard normal distribution lies to the
# LEFT of z = 2.

cdf_value = norm.cdf(z_absolute)

print("CDF value:", cdf_value)


# -----------------------------------------
# 4. What does 1 - CDF mean?
# -----------------------------------------

# CDF gives area to the LEFT.
#
# 1 - CDF gives area to the RIGHT.
#
# Therefore:
#
# 1 - 0.9772 = 0.0228
#
# So approximately 2.28% of the distribution
# lies to the RIGHT of z = 2.

right_tail = 1 - cdf_value

print("Right-tail probability:", right_tail)


# -----------------------------------------
# 5. Why multiply by 2?
# -----------------------------------------

# This is a TWO-TAILED test.
#
# We are checking:
#
# H1: μ ≠ 30
#
# So extreme values on BOTH sides matter:
#
#       Left tail       Right tail
#          ↓                ↓
#       -2.0               +2.0
#
# Each tail has approximately 0.0228 probability.
#
# Therefore:
#
# p-value = 0.0228 + 0.0228
#         = 0.0456

p_value = 2 * right_tail

print("p-value:", p_value)


# -----------------------------------------
# 6. Make the statistical decision
# -----------------------------------------

if p_value < alpha:
    print("Reject H0")
else:
    print("Fail to reject H0")
```

Problem

A company claims that the average customer waiting time is 20 minutes.

A random sample of 16 customers gives the following waiting times:

18, 22, 21, 19, 25, 17, 20, 23,
24, 18, 21, 19, 22, 20, 26, 21

At a 5% significance level, test whether the actual average waiting time is different from 20 minutes.

```python
import numpy as np
from scipy.stats import ttest_1samp


# --------------------------------------------------
# 1. Sample data
# --------------------------------------------------

waiting_times = np.array([
    18, 22, 21, 19, 25, 17, 20, 23,
    24, 18, 21, 19, 22, 20, 26, 21
])

# np.array() converts the data into a NumPy array.
# This makes it easy to perform mathematical
# operations on all observations.


# --------------------------------------------------
# 2. Hypothesized population mean
# --------------------------------------------------

population_mean = 20

# This is the value stated in the null hypothesis.
#
# H0: μ = 20
#
# H1: μ ≠ 20


# --------------------------------------------------
# 3. Significance level
# --------------------------------------------------

alpha = 0.05

# α = 0.05 means we are willing to accept
# a 5% probability of rejecting H0 when H0
# is actually true.


# --------------------------------------------------
# 4. Calculate the sample mean
# --------------------------------------------------

sample_mean = np.mean(waiting_times)

print("Sample mean:", sample_mean)

# np.mean() calculates the arithmetic mean.
#
# Sample mean = 21


# --------------------------------------------------
# 5. Perform the one-sample T-test
# --------------------------------------------------

t_statistic, p_value = ttest_1samp(
    waiting_times,
    population_mean
)

# ttest_1samp() performs a one-sample T-test.
#
# First argument:
# waiting_times
# → Our sample data
#
# Second argument:
# population_mean
# → The value assumed under H0
#
# The function returns TWO values:
#
# 1. t_statistic
# 2. p_value


# --------------------------------------------------
# 6. Display the results
# --------------------------------------------------

print("T-statistic:", t_statistic)
print("p-value:", p_value)


# --------------------------------------------------
# 7. Make the decision
# --------------------------------------------------

if p_value < alpha:
    print("Reject H0")
else:
    print("Fail to reject H0")
```
